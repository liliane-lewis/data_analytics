    To load an image and display it :

```python
from PIL import Image
import matplotlib.pyplot as plt
# Load an image using PIL
image_path = 'flowers/flowers/19_010.png'
original_image = Image.open(image_path)

# Display the original image using matplotlib
plt.imshow(original_image)
```

    Rotate an image by 90 degrees

```python

from scipy.ndimage import rotate
def rotate_image_30_degrees(image):
  return rotate(image, angle=30, reshape=True)
```

```python
rotated_array = rotate_image_30_degrees(original_image)

plt.imshow(rotated_array)
```

```python
from PIL import Image, ImageOps
def flip_image(image):
    mirrored = ImageOps.mirror(image)
    flipped = ImageOps.flip(mirrored)
    return flipped
flipped = flip_image(original_image)
plt.imshow(flipped)
```

```python
def zoom_image(image, scale=1.2):
    w, h = image.size
    new_size = (int(w * scale), int(h * scale))
    return image.resize(new_size, Image.LANCZOS)
zoomed = zoom_image(original_image, 1.2)

plt.imshow(zoomed)
```
