# **PANDAS**

**Importing Pandas**

```python
import pandas as pd
```

**Understanding Series and DataFrames**

Pandas has two primary data structures:

    Series: A one-dimensional array-like object capable of holding any data type. Think of it as a column in a spreadsheet.



```python
data = pd.Series([1, 3, 5, 7, 9])
```

DataFrames: A two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns). It’s like a spreadsheet or SQL table.

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 32],
    'City': ['New York', 'Paris', 'Berlin', 'London']
}
df = pd.DataFrame(data)
```

**Basic Operations with DataFramesLoading Data**

Pandas makes it easy to load data from various sources like CSV files, Excel spreadsheets, or SQL databases:

df = pd.read_csv('path/to/your/csvfile.csv')


**Viewing Data**

```python
df.head()  # Displays the first 5 rows by default
```

**Accessing Data** Access data in a DataFrame by referencing the column name:

```python
ages = df['Age']  # Access the 'Age' column
```

```python
print(ages)
```

**Basic Data Manipulation**

Sorting: Sort data by a specific column.

```python
df.sort_values(by='Age')
```

**Filtering: Select rows based on a certain condition.**

```python
df[df['Age'] > 30]  # Selects rows where age is greater than 30
```

**Grouping: Group data and perform aggregate functions.**

```python
df.groupby('City')['Age'].mean()  # Groups data by city and calculates mean for each group
```

# 🚀 **Challenge **

We'll simulate a dataset of a small bookstore's sales records. Our dataset will include columns for Book Title, Author, Genre, Price, and Copies Sold.

Here's how you can create this dataset using Pandas:

```python
import pandas as pd

data = {
    'Book Title': ['The Great Gatsby', 'To Kill a Mockingbird', '1984', 'Pride and Prejudice', 'The Catcher in the Rye'],
    'Author': ['F. Scott Fitzgerald', 'Harper Lee', 'George Orwell', 'Jane Austen', 'J.D. Salinger'],
    'Genre': ['Classic', 'Classic', 'Dystopian', 'Classic', 'Classic'],
    'Price': [10.99, 8.99, 7.99, 11.99, 9.99],
    'Copies Sold': [500, 600, 800, 300, 450]
}

df = pd.DataFrame(data)
```

1. Use head() to view the first few rows of the DataFrame.

```python
print(df.head())
```

2. Use describe() to get a statistical summary of the numerical columns.

```python
print(df.describe())
```

3. Use info() to get a concise summary of the DataFrame, including the number of non-null entries in each column.

```python
print(df.info())
```

4. Sort the DataFrame based on the Price or Copies Sold.

```python
print(df.sort_values(by='Copies Sold', ascending=False))
```

```python
print(df.sort_values(by='Price', ascending=False))
```

5. Filter the books by a specific Genre or books with Price above a certain threshold.

```python
print(df[df['Genre'] == 'Classic']) #
```

```python
print(df[df['Price'] > 10])
```

6. Group the books by Author and sum up the Copies Sold.

```python
print(df.groupby('Author')['Copies Sold'].sum())
```
