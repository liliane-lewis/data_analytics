````python
mean = iris_df["SepalLengthCm"].mean()
median = iris_df["SepalLengthCm"].median()
mode = iris_df["SepalLengthCm"].mode().iloc[0]

print(f"Mean: {mean}")
print(f"Median: {median}")
print(f"Mode: {mode}")
````

````python
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
plt.figure(figsize=(8,5))
sns.histplot(iris_df["SepalLengthCm"], bins=10, kde=True)
plt.title("Distribution of Sepal Length (cm)")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.grid(True)
plt.tight_layout()
plt.show()
````

#### **Interpretation:**
- The **mean sepal length** is approximately 5.84 cm, which gives us the average size of the sepal across all species.
- The **median** value (5.8 cm) is close to the mean, suggesting a relatively symmetrical distribution.
- The **mode**, being 5.0 cm, indicates that this is the most frequently occurring sepal length in the dataset.

#### **Visualization:**
A histogram was generated to show the distribution of sepal lengths. The KDE (Kernel Density Estimate) overlay helps identify the underlying shape of the data distribution, confirming that it is close to normal with a slight skew.

This basic analysis helps us understand the central tendency and distribution of one of the key flower attributes.

````python
sleep_df = pd.read_csv("Time Americans Spend Sleeping.csv")


sleep_df.head(), sleep_df.columns
````

````python
from IPython.display import Markdown, display

markdown_table = """
###Interesting Columns for Analysis — *How Much Sleep Do Americans Really Get?*

| Column                      | Possible Use Case         | Why It's Interesting                                                                 |
|----------------------------|---------------------------|---------------------------------------------------------------------------------------|
| Year                       | Trend Analysis            | Allows observation of how average sleep hours have changed over time.                |
| Avg hrs per day sleeping   | Quantitative Analysis     | The main variable to analyze by year, gender, age group, etc.                        |
| Type of Days               | Group Comparison          | Enables comparison of sleep between “All days”, “Weekdays”, and “Weekends”.          |
| Age Group                  | Demographic Comparison    | Helps explore sleep patterns across different age groups.                            |
| Sex                        | Gender-based Comparison   | Useful to study differences in average sleep time between men and women.             |
"""

display(Markdown(markdown_table))
````

````python

import matplotlib.pyplot as plt
import seaborn as sns
trend_df = sleep_df[sleep_df['Type of Days'] == 'All days']
avg_by_year = trend_df.groupby('Year')['Avg hrs per day sleeping'].mean().reset_index()

plt.figure(figsize=(10, 5))
sns.lineplot(data=avg_by_year, x='Year', y='Avg hrs per day sleeping', marker='o')
plt.title("Average Sleep Hours per Day Over Time (All Days)")
plt.xlabel("Year")
plt.ylabel("Average Hours of Sleep")
plt.grid(True)
plt.tight_layout()
plt.show()
````

````python
avg_by_sex = sleep_df.groupby('Sex')['Avg hrs per day sleeping'].mean().reset_index()

plt.figure(figsize=(6, 4))
sns.barplot(data=avg_by_sex, x='Sex', y='Avg hrs per day sleeping')
plt.title("Average Sleep Hours per Day by Sex")
plt.xlabel("Sex")
plt.ylabel("Average Hours of Sleep")
plt.tight_layout()
plt.show()
````
